```python
import numpy as np
from sympleq.core.paulis import PauliSum
from sympleq.applications.measurement.aquire import AquireConfig, Aquire
```

```python
# Hamiltonian
dims = [2, 2]
paulis = ["x0z1 x0z1", "x1z0 x1z1", "x0z0 x0z1"]
coefficients = [1., 1., 1.]
P = PauliSum.from_string(paulis, dims, coefficients)
```

```python
# State
psi = np.random.rand(int(np.prod(dims))) + 1j * np.random.rand(int(np.prod(dims)))
psi = psi / np.linalg.norm(psi)
```

```python
print(P)
print(psi)
```

```python
# Update steps
update_steps = [25, 50, 100, 200, 400, 800, 1600, 3200, 6400, 12800, 25600, 51200, 102400]
```

```python
# AQUIRE
model = Aquire(H=P, psi=psi)
```

```python
print(model)
```
