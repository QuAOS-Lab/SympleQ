## Create and compare Pauli strings

Pauli strings can be parsed from their text representation or built from their X and Z exponent vectors. Here the first string is X and the second is Z, so they do not commute.

```python
from sympleq_light import PauliString, PauliSum

x = PauliString.from_string("x1z0")
z = PauliString.from_exponents([0], [1])

print("X:", x)
print("Z:", z)
print("Do they commute?", x.commutes_with(z))
assert not x.commutes_with(z)
```

## Create a zero Pauli sum

Pass the qudit dimensions to `PauliSum.zero`. Dimensions must be prime; this example uses a qubit and a qutrit.

```python
zero = PauliSum.zero([2, 3])
print(zero)
```

Pauli-string text uses `x<exponent>z<exponent>` tokens. For non-qubit qudits, append `/dimension` to the token. A Pauli sum can be parsed from the textual format produced by `str(sum)`.
